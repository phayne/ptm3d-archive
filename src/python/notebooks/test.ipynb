```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import time series data from PTM3D output
file1 = '/Users/paha3326/Library/CloudStorage/OneDrive-UCB-O365/europa/heat3d/ptm/output/moon8_global_heat-flow-low/ptm3d_plot_data_HAWORTH-CENTER_heatflow-low.dat'
file2 = '/Users/paha3326/Library/CloudStorage/OneDrive-UCB-O365/europa/heat3d/ptm/output/moon8_global_heat-flow-high/ptm3d_plot_data_HAWORTH-CENTER_heatflow-high.dat'

# Columns: 0=Timestep, 1=Time_Label(string), 2=Time_d, 3=Temperature_K
# Skip column 1 (ISO datetime string) by selecting only numeric columns
data1 = np.loadtxt(file1, skiprows=6, usecols=(0, 2, 3))
data2 = np.loadtxt(file2, skiprows=6, usecols=(0, 2, 3))
time1 = data1[:, 1]  # Time_d (days)
time2 = data2[:, 1]  # Time_d (days)
tsurf1 = data1[:, 2]  # Temperature_K
tsurf2 = data2[:, 2]  # Temperature_K

# Plot surface temperature vs time
plt.figure(figsize=(10, 6))
plt.plot(time1, tsurf1, label='Low Heat Flow', color='blue')
plt.plot(time2, tsurf2, label='High Heat Flow', color='red')
plt.xlabel('Time (days)')
plt.ylabel('Surface Temperature (K)')
plt.title('Surface Temperature vs Time at Haworth Center')
plt.legend(frameon=False)
```
